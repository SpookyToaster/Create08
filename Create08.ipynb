```sql
--Create transactions

* Animals
  * What is the animal
  * What is it’s name
  * How old is it
  * Where is it located
  * What does it eat
  * Who is it’s trainer
* Enclosures
  * What is the enclosure (name)
 What is the maximum and minimum temp of the enclosure (in
Fahrenheit)
  * What is the square footage of the enclosure
  * Which groundskeeper is responsible for the enclosure
* Employees
  * What is the employees name
  * What is their job category
  * What is their salary
  * Which shift are they? (day/night)


-- this layout seems to be a good start point. Next will review the query requirements to see if they look generally applicable to the current schema. 
-- will leave this layout here as a way to create the ERD chart later on.

-- tables needed
cc_animal
-- animal_id (primary key)
-- animal_name
-- animal_type
-- animal_age
-- animal_location (enclosure_id foreign key)
-- trainer_id (foreign key with cc_employee table?)

cc_enclosure
-- enclosure_id (primary key)
-- enclosure_name
-- max_temp F
-- min_temp F
-- square_footage
-- groundskeeper_id

cc_employee
-- employee_id (primary key)
-- employee_name
-- job_category
-- salary
-- shift (day/night)


-- BEGIN PROGRAM

BEGIN TRANSACTION;
  BEGIN TRY
    -- create the tables
    CREATE TABLE cc_animal (
        animal_id INT PRIMARY KEY,
        -- adjust length after we do transactions and see what the longest name is? or a reasonable length for the name of the animal
        animal_name VARCHAR(100),
        -- same here as above
        animal_type VARCHAR(50),
        animal_age INT,
        animal_location INT,
        trainer_id INT,
        -- adding foreign keys to reference the enclosure and employee tables
        FOREIGN KEY (animal_location) REFERENCES cc_enclosure(enclosure_id),
        -- creating FK with E_id. Assumption is that the employee Job_category will be trainer, but will need to check that in the transactions.
        FOREIGN KEY (trainer_id) REFERENCES cc_employee(employee_id)
    );

    CREATE TABLE cc_enclosure (
        enclosure_id INT PRIMARY KEY,
        -- again, check length for reasonable length for the name of the enclosure
        enclosure_name VARCHAR(100),
        -- ensure no entry by celcius, otherwise need a way to handle
        max_temp INT,
        min_temp INT,
        square_footage INT,
        groundskeeper_id INT,
        -- creating FK with E_id. Assumption is that the employee Job_category will be groundskeeper, but will need to check that in the transactions. so no need to create a separate table for groundskeeper, just use the employee table and check the job_category. Same with trainer i think.
        FOREIGN KEY (groundskeeper_id) REFERENCES cc_employee(employee_id)
    );

    CREATE TABLE cc_employee (
        employee_id INT PRIMARY KEY,
        -- check length for reasonable length for the name of the employee
        employee_name VARCHAR(100),
        -- check length for reasonable length for the job category of the employee
        job_category VARCHAR(50),
        salary DECIMAL(10, 2),
        shift VARCHAR(10)
    );

  END TRY
  
    BEGIN CATCH
      -- Rollback the transaction if an error occurs
      ROLLBACK TRANSACTION;
      -- Actually catch the error message, not just a generic message saying an error occurred. This will help with debugging.
      DECLARE @ErrorMessage NVARCHAR(4000);  
    END CATCH;
    
END TRANSACTION;

```
